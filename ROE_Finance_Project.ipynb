```python
def get_bank_data(ticker):

    company = yf.Ticker(ticker)

    revenue = company.financials.loc["Total Revenue"].iloc[0]

    net_income = company.financials.loc["Net Income"].iloc[0]

    assets = company.balance_sheet.loc["Total Assets"].iloc[0]

    equity = company.balance_sheet.loc["Stockholders Equity"].iloc[0]

    roe = net_income / equity

    return {
        "Company": ticker,
        "Revenue": revenue,
        "Net Income": net_income,
        "Assets": assets,
        "Equity": equity,
        "ROE": roe
    }
```

```python
jpm_data = get_bank_data("JPM")

jpm_data
```

```python
banks = ["JPM", "BAC", "WFC", "GS", "MS"]

data = []

for ticker in banks:
    data.append(get_bank_data(ticker))

df = pd.DataFrame(data)

df
```

```python
df["ROE"] = df["ROE"] * 100

df
```

```python
df.sort_values("ROE", ascending=False)
```

```python
def get_stock_return(ticker):

    stock = yf.Ticker(ticker)

    prices = stock.history(period="1y")

    start_price = prices["Close"].iloc[0]

    end_price = prices["Close"].iloc[-1]

    return ((end_price - start_price) / start_price) * 100
```

```python
get_stock_return("JPM")
```

```python
df["Stock Return"] = df["Company"].apply(get_stock_return)

df
```

```python
def get_bank_data(ticker):

    company = yf.Ticker(ticker)

    revenue = company.financials.loc["Total Revenue"].iloc[0]
    previous_revenue = company.financials.loc["Total Revenue"].iloc[1]

    revenue_growth = ((revenue - previous_revenue) / previous_revenue) * 100

    net_income = company.financials.loc["Net Income"].iloc[0]

    assets = company.balance_sheet.loc["Total Assets"].iloc[0]

    equity = company.balance_sheet.loc["Stockholders Equity"].iloc[0]

    roe = (net_income / equity) * 100

    return {
        "Company": ticker,
        "Revenue": revenue,
        "Revenue Growth": revenue_growth,
        "Net Income": net_income,
        "Assets": assets,
        "Equity": equity,
        "ROE": roe
    }
```

```python
banks = ["JPM", "BAC", "WFC", "GS", "MS"]

data = []

for ticker in banks:
    data.append(get_bank_data(ticker))

df = pd.DataFrame(data)

df
```

```python
df["Stock Return"] = df["Company"].apply(get_stock_return)

df
```

```python
corr_growth = df["Revenue Growth"].corr(df["Stock Return"])

print(corr_growth)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.scatter(df["Revenue Growth"], df["Stock Return"])

for i, company in enumerate(df["Company"]):
    plt.annotate(
        company,
        (
            df["Revenue Growth"].iloc[i],
            df["Stock Return"].iloc[i]
        )
    )

plt.xlabel("Revenue Growth (%)")
plt.ylabel("Stock Return (%)")
plt.title("Revenue Growth vs Stock Return")

plt.show()
```
